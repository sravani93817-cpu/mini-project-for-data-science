```python
# --- GLOBAL SAMPLE DATA ---
# This dictionary represents our store inventory database
STORE_INVENTORY = {
    1: {"name": "Laptop", "price": 899.99, "stock": 5},
    2: {"name": "Wireless Mouse", "price": 25.50, "stock": 15},
    3: {"name": "Gaming Headphones", "price": 79.00, "stock": 8},
    4: {"name": "USB-C Cable", "price": 12.99, "stock": 0}  # Out of stock item
}

class ECommerceStore:
    def __init__(self):
        """Initializes an empty shopping cart for the user."""
        # Format: {item_id: selected_quantity}
        self.cart = {}

    def display_products(self):
        """Prints all products in the database cleanly."""
        print("\n=== WELCOME TO THE PYTHON MINI-STORE ===")
        print(f"{'ID':<5} {'Product Name':<20} {'Price':<10} {'Availability':<12}")
        print("-" * 52)

        for item_id, details in STORE_INVENTORY.items():
            status = f"{details['stock']} left" if details['stock'] > 0 else "OUT OF STOCK"
            print(f"{item_id:<5} {details['name']:<20} ${details['price']:<10.2f} {status:<12}")
        print("-" * 52)

    def add_to_cart(self):
        """Handles selecting items and checking if stock is available."""
        try:
            item_id = int(input("Enter the Product ID you want to buy: "))

            # Check if product exists
            if item_id not in STORE_INVENTORY:
                print("❌ Invalid Product ID! Please try again.\n")
                return

            product = STORE_INVENTORY[item_id]

            # Check if product is out of stock
            if product["stock"] <= 0:
                print("❌ Sorry, this item is currently out of stock!\n")
                return

            quantity = int(input(f"How many '{product['name']}' would you like?: "))

            # Check if user requested more than available stock
            if quantity > product["stock"]:
                print(f"❌ Error: We only have {product['stock']} units available.\n")
                return
            if quantity <= 0:
                print("❌ Quantity must be at least 1.\n")
                return

            # Update inventory and add to cart
            product["stock"] -= quantity
            self.cart[item_id] = self.cart.get(item_id, 0) + quantity
            print(f"🛒 Successfully added {quantity}x '{product['name']}' to your cart!\n")

        except ValueError:
            print("❌ Invalid input! Please type numbers only.\n")

    def view_cart_and_checkout(self):
        """Displays user cart items, computes pricing totals, and clears cart."""
        if not self.cart:
            print("🛒 Your shopping cart is empty!\n")
            return

        print("\n=== YOUR SHOPPING CART ===")
        subtotal = 0.0

        for item_id, qty in self.cart.items():
            product = STORE_INVENTORY[item_id]
            cost = product["price"] * qty
            subtotal += cost
            print(f"• {product['name']} (x{qty}): ${cost:.2f}")

        print("-" * 26)
        print(f"Subtotal: ${subtotal:.2f}")

        # Simple Business Rule: 10% discount if spending is over $100
        discount = 0.0
        if subtotal > 100:
            discount = subtotal * 0.10
            print(f"🎉 Order over $100! 10% Discount: -${discount:.2f}")

        final_total = subtotal - discount
        print(f"Total Bill Amount: ${final_total:.2f}")
        print("==========================\n")

        # Complete checkout simulation
        confirm = input("Type 'BUY' to confirm purchase or any key to return: ").strip().upper()
        if confirm == "BUY":
            self.cart.clear() # Empty the cart after successful purchase
            print("✅ Order placed successfully! Thank you for shopping with us.\n")
        else:
            print("⚠️ Checkout canceled. Items remain in your cart.\n")


def main():
    """Main terminal loop system."""
    store = ECommerceStore()

    while True:
        print("=== STORE NAVIGATION MENU ===")
        print("1. View Store Products")
        print("2. Add Product to Cart")
        print("3. View Cart & Checkout")
        print("4. Exit Store")

        choice = input("Select an option (1-4): ").strip()

        if choice == "1":
            store.display_products()
        elif choice == "2":
            store.display_products() # Show options before asking
            store.add_to_cart()
        elif choice == "3":
            store.view_cart_and_checkout()
        elif choice == "4":
            print("Thank you for visiting! Goodbye.")
            break
        else:
            print("❌ Invalid option choice. Please type 1, 2, 3, or 4.\n")

if __name__ == "__main__":
    main()
```
